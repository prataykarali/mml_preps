```python
import torch
import math

# Taylor center
x0 = 0.0

# Autodiff variable
x = torch.tensor(x0, requires_grad=True)

# Function
f = torch.sin(x) + torch.cos(x)

# Store derivatives
derivatives = [f]

current = f

for k in range(1, 6):
    current = torch.autograd.grad(
        current,
        x,
        create_graph=(k < 5)
    )[0]

    derivatives.append(current)

# Taylor coefficients
for k, derivative in enumerate(derivatives):

    coefficient = derivative.item() / math.factorial(k)

    print(
        f"T{k}: {coefficient:.4f} * "
        f"(x - {x0})^{k}"
    )
```
