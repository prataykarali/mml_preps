```python
import torch
from torch.nn import functional as F
def f1(x):
    return torch.sin(x[0])*torch.cos(x[1])
x=torch.tensor([2.0,3.0],requires_grad=True)
J1=torch.autograd.functional.jacobian(f1,x)
print("Jacobian of f1:", J1)
print("df1/dx1:", J1[0])

def f2(x,y):
    return torch.dot(x,y)
x=torch.tensor([1.0,2.0],requires_grad=True)
y=torch.tensor([3.0,4.0],requires_grad=True)
J2=torch.autograd.functional.jacobian(f2,(x,y))

print("Jacobian of f2:", J2)

def f3(x):
    return torch.outer(x,x)
x=torch.tensor([1.0,2.0],requires_grad=True)
J3=torch.autograd.functional.jacobian(f3,x)

print("Jacobian of f3:", J3)
```
