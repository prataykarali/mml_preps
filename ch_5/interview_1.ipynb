```python
import torch
from torch.nn import functional as F
z=torch.tensor(2.0, requires_grad=True)
y=F.softplus(z)
y.backward()
print("z =", z.item())
print("Softplus(z) =", y.item())
print("d Softplus/dz =", z.grad.item())

# Standard sigmoid
print("Sigmoid(z) =", torch.sigmoid(z).item())
```
